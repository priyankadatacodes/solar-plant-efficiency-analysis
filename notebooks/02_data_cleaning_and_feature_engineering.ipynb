```python
import pandas as pd
import numpy as np
```

```python
solar_df = pd.read_csv(
    "solar_plant_cleaned_data.csv",
    parse_dates=['date_time']
)
```

```python
solar_df['date'] = solar_df['date_time'].dt.date
solar_df['month'] = solar_df['date_time'].dt.to_period('M')
```

```python
solar_df['date'] = solar_df['date_time'].dt.date
solar_df['month'] = solar_df['date_time'].dt.to_period('M')
```

```python
panel_efficiency = (
    solar_df
    .groupby('inverter_id')['efficiency_pct']
    .mean()
    .reset_index()
    .sort_values(by='efficiency_pct', ascending=False)
)
```

```python
solar_df['energy_loss_pct'] = np.where(
    solar_df['dc_power'] > 0,
    (solar_df['power_loss_kw'] / solar_df['dc_power']) * 100,
    np.nan
)
```

```python
energy_loss_summary = solar_df['energy_loss_pct'].mean()
energy_loss_summary
```

```python
daily_yield = (
    solar_df
    .groupby('date')['daily_yield']
    .max()
    .reset_index()
)
```

```python
monthly_yield = (
    solar_df
    .groupby('month')['daily_yield']
    .max()
    .reset_index()
)
```

```python
specific_yield_kpi = (
    solar_df
    .groupby('inverter_id')['specific_yield']
    .mean()
    .reset_index()
    .sort_values(by='specific_yield')
)
```

```python
low_performing_assets = panel_efficiency[
    panel_efficiency['efficiency_pct'] <
    panel_efficiency['efficiency_pct'].mean()
]
```

```python
executive_kpis = {
    "Average Panel Efficiency (%)": round(panel_efficiency['efficiency_pct'].mean(), 2),
    "Average Energy Loss (%)": round(energy_loss_summary, 2),
    "Total Monthly Yield (Latest)": monthly_yield.iloc[-1]['daily_yield']
}

executive_kpis
```
